```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

text = "Sistem temu kembali informasi adalah teknologi yang sangat penting dalam pencarian dokumen"

tokens = word_tokenize(text.lower())

stop_words = set(stopwords.words('indonesian'))
filtered_tokens = [word for word in tokens if word not in stop_words]

print('Teks Asli:', text)
print('Token Setelah Stopword Removal:', filtered_tokens)
```

```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()
words = ['running', 'runs', 'easily', 'faster']

stemmed_words = [ps.stem(words) for words in words]
print(stemmed_words)
```

```python
import nltk
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download('stopwords')

# Teks artikel
text = 'Sistem temu kembali informasi sangat berguna dalam pencarian data dan dokumen. Penggunaan teknologi ini memungkinkan akses cepat dan mudah.'

# 1. Caase Foldling
text = text.lower()

# 2. Tekonisasi
tokens = word_tokenize(text)

# 3. Stopword Removal
stop_words = set(stopwords.words('indonesian'))
filtered_tokens = [word for word in tokens if word not in stop_words and word.isalnum()]

# 4. Stemming
factory = StemmerFactory()
stemmer = factory.create_stemmer()
stemmed_tokens = [stemmer.stem(word) for word in filtered_tokens]

# Hasil Akhir
print('Teks Asli:', text)
print('')
print('Token Setelah Stopword Removal dan Stemming:', stemmed_tokens)
```

```python
import pandas as pd
import numpy as np
from nltk.corpus import stopwords
import nltk
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
```

```python
df = pd.read_excel('datasetweek4stk.xlsx', header=0)

df.head()
```

```python
def clean_lower(lwr):
    lwr = lwr.lower()
    return lwr
```

```python
df['hasil_casefolding'] = df['data'].apply(clean_lower)
casefolding = pd.DataFrame(df['hasil_casefolding'])
casefolding
```

```python
stopword = set(stopwords.words('indonesian'))

def clean_stopwords(text):
    text = ' '.join(word for word in text.split() if word not in stopword)
    return text

df['hasil_stopword'] = df['hasil_casefolding'].apply(clean_stopwords)
```

```python
stopwordremoval = pd.DataFrame(df['hasil_stopword'])

stopwordremoval
```

```python
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

def get_stemmed_term(text):
    factory = StemmerFactory()
    stemmer = factory.create_stemmer()
    hasil = stemmer.stem(text)
    return hasil

df['hasil_stemming'] = df['hasil_stopword'].apply(get_stemmed_term)
```

```python
stemmingSastrawi = pd.DataFrame(df['hasil_stemming'])
stemmingSastrawi
```

```python
from nltk.tokenize import word_tokenize
def word_tokens(text):
    token = word_tokenize(text)
    return token

df['hasil_tokenisasi'] = df['hasil_stemming'].apply(word_tokens)
```

```python
tokenkata = pd.DataFrame(df['hasil_tokenisasi'])
tokenkata
```

```python
teks_Asli = pd.Series(df['data'])
casefolding = pd.Series(df['hasil_casefolding'])
stopwordremoval = pd.Series(df['hasil_stopword'])
stemming = pd.Series(df['hasil_stemming'])
tokenisasi = pd.Series(df['hasil_tokenisasi'])

df = pd.DataFrame({
    'Teks Asli': teks_Asli,
    'Hasil Casefolding': casefolding,
    'Hasil Stopword Removal': stopwordremoval,
    'Hasil Stem': stemming,
    'Hasil Tokenisasi': tokenisasi
})
```

```python
hasil_pre_pro = df.to_excel('Hasil Pre-Processing Week 4 STK.xlsx', sheet_name='sheet_satu')
```
