```python
import pandas as pd
import nltk
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

nltk.download('punkt')
nltk.download('stopwords')
```

```python
df = pd.read_excel('dataset_tugas_week4_stk.xlsx')
df
```

```python
def casefolding(text):
    return text.lower()

df['Hasil Casefolding'] = df['Komentar'].apply(casefolding)
hasil = pd.DataFrame(df['Hasil Casefolding'])

hasil
```

```python
stopwords = set(stopwords.words('indonesian'))

tambahan_stopword = ['yg', 'd', 'di', 'ke', 'dari', 'klw', 'ga', 'jd', 'kl', 'gk', 'tpi', 'smw', 'udh']
stopwords.update(tambahan_stopword)

def clean_stopwords(text):
    text = ' '.join(word for word in text.split() if word not in stopwords)
    return text

df['Hasil Stopword'] = df['Hasil Casefolding'].apply(clean_stopwords)

stopwordremoval = pd.DataFrame(df['Hasil Stopword'])
stopwordremoval
```

```python
def get_stemmed_term(text):
    factory = StemmerFactory()
    stemmer = factory.create_stemmer()
    hasil = stemmer.stem(text)
    return hasil

df['Hasil Stemming'] = df['Hasil Stopword'].apply(get_stemmed_term)
stemmingSastrawi = pd.DataFrame(df['Hasil Stemming'])

stemmingSastrawi
```

```python
def word_tokens(text):
    token = word_tokenize(text)
    return token

df['Hasil Tokenisasi'] = df['Hasil Stemming'].apply(word_tokens)
tokenkata = pd.DataFrame(df['Hasil Tokenisasi'])

tokenkata
```

```python
komen_asli = pd.Series(df['Komentar'])
casefolding = pd.Series(df['Hasil Casefolding'])
stopwordremoval = pd.Series(df['Hasil Stopword'])
stemming = pd.Series(df['Hasil Stemming'])
tokenisasi = pd.Series(df['Hasil Tokenisasi'])

df = pd.DataFrame({
    'Komentar Asli': komen_asli,
    'Hasil Casefolding': casefolding,
    'Hasil Stopword Removal': stopwordremoval,
    'Hasil Stem': stemming,
    'Hasil Tokenisasi': tokenisasi
})

df.to_excel('Hasil Prapemrosesan.xlsx', index=False)
```

```python
df_skenario = pd.DataFrame()
df_skenario['Komentar Asli'] = df['Komentar Asli']

df_skenario['Skenario_Casefolding'] = df_skenario['Komentar Asli'].str.lower()

df_skenario['Skenario_Stemming'] = df_skenario['Skenario_Casefolding'].apply(get_stemmed_term)

df_skenario['Skenario_Stopword'] = df_skenario['Skenario_Stemming'].apply(clean_stopwords)

df_skenario['Skenario_Tokenisasi'] = df_skenario['Skenario_Stopword'].apply(word_tokens)

df_perbandingan = pd.DataFrame({
    'Komentar Asli': df['Komentar Asli'],
    'Tokenisasi (Stopword lalu Stemming)': df['Hasil Tokenisasi'],
    'Tokenisasi (Stemming lalu Stopword)': df_skenario['Skenario_Tokenisasi']
})

df_perbandingan
```
