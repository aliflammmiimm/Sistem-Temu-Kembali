```python
# Membuat array 1 dimensi
array_1d = np.array([1, 2, 3, 4, 5])
print("Array 1D:", array_1d)

# Membuat array 2 dimensi
array_2d = np.array([[1, 2, 3], [4, 5, 6]])
print("Array 2D:\n", array_2d)

# Mengakses elemen array
print("Elemen pertama array_1d:", array_1d[0])
print("Elemen di baris ke-2, kolom ke-3 dari array_2d:", array_2d[1, 2])

# Operasi dasar pada array
print("Penjumlahan array:", array_1d + 5)
print("Perkalian array:", array_1d * 2)

# Fungsi matematika
print("Akar kuadrat dari array_1d:", np.sqrt(array_1d))
print("Sine dari array_1d:", np.sin(array_1d))

# Reshape array
array_reshaped = array_1d.reshape((1, 5))
print("Array yang telah di reshape:\n", array_reshaped)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

# Dummy dataset
n = 100

df = pd.DataFrame({
    "article_id": range(1, n+1),
    "category": np.random.choice(["Politik", "Ekonomi", "Olahraga"], n),
    "word_count": np.random.randint(200,2000,n),
    "views": np.random.randint(100,10000,n),
    "likes": np.random.randint(10,2000,n),
    "comments": np.random.randint(0,500,n)
})

df.head()
```

```python
plt.figure()
sns.histplot(data=df, x="word_count", bins=20, kde=True)
plt.title("Distribusi Panjang Artikel (word count)")
plt.xlabel("Jumlah Kata")
plt.ylabel("Frekuensi")
plt.show()
```

```python
plt.figure()
sns.countplot(data=df, x="category")
plt.title("Jumlah Artikel per Kategori")
plt.xlabel("Kategori")
plt.ylabel("Jumlah Artikel")
plt.show()
```

```python
import numpy as np

plt.figure()
sns.barplot(data=df, x="category", y="views", estimator=np.mean, errorbar=('ci', 95))

plt.title("Rata-rata Views per Kategori (95% CI)")
plt.xlabel("Kategori")
plt.ylabel("Rata-rata Views")
plt.show()
```

```python
plt.figure()
sns.boxplot(data=df, x="category", y="word_count")

plt.title("Sebaran Panjang Artikel per Kategori (Boxplot)")
plt.xlabel("Kategori")
plt.ylabel("Jumlah Kata")
plt.show()
```

```python
plt.figure()
sns.scatterplot(data=df, x="word_count", y="views", hue="category")

plt.title("Panjang Artikel vs Views per Kategori")
plt.xlabel("Jumlah kata")
plt.ylabel("Views")
plt.show()
```

```python
num_cols = ["word_count", "views", "likes", "comments"]
corr = df[num_cols].corr(numeric_only=True)

plt.figure()
sns.heatmap(corr, annot=True, fmt=".2f", cmap="Blues")
plt.title("Korelasi Antar Variabel Numerik")
plt.show()
```

```python
import numpy as np
from scipy import stats

data = [65, 70, 70, 80, 85, 90, 90, 90, 95, 100]

mean_val = np.mean(data)

median_val = np.median(data)

mode_val = stats.mode(data, keepdims=True)
mode_num = mode_val.mode.item()
mode_count = mode_val.count.item()

std_val = np.std(data)
var_val = np.var(data)

print("Data:", data)
print("Mean:", mean_val)
print("Median:", median_val)
print("Mode:", mode_num, "(muncul", mode_count, "kali)")
print("Standard Deviasi:", round(std_val, 2))
print("Variance:", round(var_val, 2))
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
corpus =[
    "mesin pencari adalah sistem temu kembali informasi",
    "informasi dapat ditemukan dengan sistem pencari",
    "model vector space digunakan dalam information retrieval"
]
```

```python
cv = CountVectorizer()

X_count = cv.fit_transform(corpus)

print ("Ukuruan Matriks", X_count.shape)
print ("Daftar Kata (vocab)", cv.get_feature_names_out())
print ("Ukuruan frekuensi", X_count.toarray())
```
