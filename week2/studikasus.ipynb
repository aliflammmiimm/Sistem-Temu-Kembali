```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("BigmacPrice.csv")

df
```

```python
df.shape
```

```python
df.columns
```

```python
data_indonesia = df[df["name"] == "Indonesia"]

data_indonesia
```

```python
data_indonesia.describe()
```

```python
harga_murah = data_indonesia[data_indonesia["local_price"] < 20000]

harga_murah
```

```python
harga_mahal = data_indonesia[data_indonesia["local_price"] >= 20000]

harga_mahal
```

```python
plt.plot(harga_murah["date"], harga_murah["local_price"])
plt.xlabel("Periode")
plt.ylabel("Harga")
plt.xticks(rotation='vertical')
plt.title("Harga McD Kategori Murah")
plt.show()
```

```python
plt.plot(harga_mahal['date'], harga_mahal['local_price'])
plt.xlabel("Periode")
plt.ylabel("Harga")
plt.xticks(rotation='vertical')
plt.title("Harga McD Kategori Mahal")
plt.show()
```

```python
plt.plot(harga_murah['date'], harga_murah['local_price'])
plt.plot(harga_mahal['date'], harga_mahal['local_price'])

plt.xticks(rotation='vertical')
plt.xlabel("Periode")
plt.ylabel("Harga")
plt.title("Perbandingan Harga McD")
plt.legend(["Murah", "Mahal"])
plt.show()
```

```python
df = pd.read_csv("sample_data_plot.csv")

df.head()
```

```python
df.rename(columns={
    'Country name': 'Country',
    'Perceptions of corruption': 'Corruption',
    'Freedom to make life choices': 'Freedom'
}, inplace=True)

df.columns
```

```python
df[:5].plot(x='Country',
            y=['Corruption', 'Freedom', 'Generosity', 'Social support'],
            kind='bar')

plt.title("Contoh Bar Plot")
plt.ylabel("Nilai")
plt.show()
```

```python
df[:5].plot(x='Country',
            y=['Corruption', 'Freedom', 'Generosity', 'Social support'],
            kind='line')

plt.title("Contoh Line Plot")
plt.ylabel("Nilai")
plt.show()
```

```python
df.plot(y=['Corruption', 'Freedom', 'Generosity', 'Social support'],
            kind='box')

plt.title("Contoh Box Plot")
plt.xlabel("Parameter")
plt.ylabel("Nilai")
plt.show()
```

```python
df.plot(x='Corruption', y='Freedom', kind='scatter')

plt.xlim((0, 1))
plt.ylim((0, 1))

plt.title("Contoh Scatter Plot")
plt.show()
```

```python
df.plot(x='Country', y='Corruption', kind='hist', bins=10)

plt.title("Contoh Histogram")
plt.xlabel("Nilai")
plt.show()
```

```python
df.plot(x='Country',
        y=['Corruption', 'Freedom', 'Generosity', 'Social support'],
        kind='hist',
        subplots=True,
        layout=(2, 2))

plt.suptitle("Contoh Subplots")
plt.tight_layout()
plt.show()
```
