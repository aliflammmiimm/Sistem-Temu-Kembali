```python
plt.figure(figsize=(8,5))
plt.hist(df['total_streams_billions'], bins=10)

plt.title('Distribusi Total Streams Lagu')
plt.xlabel('Total Streams (Billions)')
plt.ylabel('Jumlah Lagu')

plt.show()
```

```python
top10 = df.sort_values(by='total_streams_billions', ascending=False).head(10)

plt.figure(figsize=(10,5))
plt.bar(top10['song_title'], top10['total_streams_billions'])

plt.title('10 Lagu dengan Total Streams Tertinggi')
plt.xlabel('Judul Lagu')
plt.ylabel('Total Streams (Billions)')
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(data=df, x='primary_genre')

plt.title('Jumlah Lagu Berdasarkan Genre')
plt.xlabel('Genre')
plt.ylabel('Jumlah Lagu')
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['danceability'], df['energy'])

plt.title('Hubungan Danceability dan Energy')
plt.xlabel('Danceability')
plt.ylabel('Energy')

plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(y=df['bpm'])

plt.title('Distribusi BPM Lagu')
plt.ylabel('BPM')

plt.show()
```
