```python
import re

statement = 'Please contact us at: support@datacamp.com'
match = re.search(r'([\w\.-]+)@([\w\.-]+)', statement)

if statement:
    print('Email address:', match.group())
    print('Username:', match.group(1))
    print('Host:', match.group(2))
```

```python
import re

tweets = [
    '#nowlistening Not You by Alan Walker & Emma Steinbakken',
    '#nowlistening Vampire by Olivia Rodrigo',
    '#nowlistening Barbie World by Nicki Minaj & Ice Spice With Aqua'
]

pattern = r'(?i)#nowlistening\s+(.*?)(?:\s+by\s+(.*))?$'

for tweet in tweets:
    match = re.search(pattern, tweet)
    if match:
        judul = match.group(1)
        penyanyi = match.group(2)
        print(f'Judul: {judul.strip()}')
        if penyanyi:
            print(f'Penyanyi: {penyanyi.strip()}')
        else:
            print('Penyanyi: Tidak Diketahui')
    else:
        print('TIdak ada hasil')
```

```python
import re

teks = '''
Saya membeli 3 buku di toko A.
Harga 1 bukunya Rp50.000,- !!!
Kunjungi website https://tokobuku.com sekarang.
Email saya: contoh123@gmail.com
Produk ini Sangat Bagus sekali!!!
Diskon besar 70% hanya hari ini
Hubungi admin di admin@shop.co.id
Saya suka Belanja Online di Tokopedia.
12345 ini angka yang harus dihapus.
Coba cek di www.contohweb.id sekarang juga!
'''

print('=== TEKS ASLI ===')
print(teks)
```

```python
teks_no_num = re.sub(r'\d+', '', teks)
print('\n[Hapus Angka]\n', teks_no_num)

teks_no_punc = re.sub(r'[^\w\s]', '', teks_no_num)
print('\n[Hapus Tanda Baca]\n', teks_no_punc)

teks_no_url = re.sub(r'http\S+|www\.\S+', '', teks_no_punc)
print('\n[Hapus URL]\n', teks_no_url)

teks_no_email = re.sub(r'\S+@\S+', '', teks_no_url)
print('\n[Hapus Email]\n', teks_no_email)

teks_bersih = re.sub(r'\s+', ' ', teks_no_email).strip()
print('\n=== TEKS BERSIH ===\n', teks_bersih)
```
