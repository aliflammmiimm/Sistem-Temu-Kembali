```python
import pandas as pd

pd.set_option('display.max_colwidth', None)
df = pd.read_csv('dataset_tugas_regex.csv')
df.head()
```

```python
import re

df['Hapus Angka'] = df['Kalimat Asli'].apply(
    lambda x: re.sub(r'\d+', '', str(x))
)

display(df[['Hapus Angka']].head(20))
```

```python
df['Hapus Tanda Baca'] = df['Hapus Angka'].apply(
    lambda x: re.sub(r'[^\w\s]', '', str(x))
)

display(df[['Hapus Tanda Baca']].tail(20))
```

```python
df['Hapus URL'] = df['Hapus Tanda Baca'].apply(
    lambda x: re.sub(r'http\S+|www\.\S+', '', str(x))
)

display(df[['Hapus URL']].head(20))
```

```python
df['Teks Bersih'] = df['Hapus URL'].apply(
    lambda x: re.sub(r'\s+', ' ', str(x)).strip()
)

display(df[['Teks Bersih']].head(20))
```

```python
df['Kata Kapital'] = df['Kalimat Asli'].apply(
    lambda x: re.findall(r'\b[A-Z][a-zA-Z]*\b', str(x))
)

display(df[['Kata Kapital']].head(20))
```

```python
df['Kata Lebih 5 Huruf'] = df['Kalimat Asli'].apply(
    lambda x: re.findall(r'\b\w{6,}\b', str(x))
)

display(df[['Kata Lebih 5 Huruf']].head(20))
```
