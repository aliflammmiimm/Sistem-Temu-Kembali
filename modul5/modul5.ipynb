```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import re
from nltk.corpus import stopwords
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

# Contoh dataset pertanyaan
questions = [
    "Bagaimana cara reset password akun saya?",
    "Apa itu machine learning?",
    "Bagaimana cara upgrade paket internet?",
    "Bagaimana cara menghubungkan printer ke wifi?"
]

# Fungsi preprocessing
factory = StemmerFactory()
stemmer = factory.create_stemmer()

def preprocess(text):
    text = text.lower()
    text = re.sub(r'\d+', '', text)
    text = re.sub(r'[^\w\s]', '', text)
    words = text.split()
    stop_words = set(stopwords.words('indonesian'))
    words = [word for word in words if word not in stop_words]
    words = [stemmer.stem(word) for word in words]
    return ' '.join(words)

# Inisialisasi TF-IDF Vectorizer
vectorizer = TfidfVectorizer()

# Mengubah pertanyaan menjadi matriks TF-IDF
tfidf_matrix = vectorizer.fit_transform(questions)

# Menentukan similarity threshold
similarity_threshold = 0.2

# Fungsi untuk menangani input pengguna
def handle_user_query(user_query):
    # Memproses query pengguna
    query_processed = preprocess(user_query)
    
    # Menghitung TF-IDF untuk query pengguna
    query_tfidf = vectorizer.transform([query_processed])
    
    # Menghitung cosine similarity
    similarity_scores = cosine_similarity(query_tfidf, tfidf_matrix)
    
    # Mengambil indeks dengan skor kemiripan tertinggi
    top_index = np.argmax(similarity_scores)
    
    # Cek apakah skor kemiripan di atas threshold
    if similarity_scores[0, top_index] < similarity_threshold:
        print("Chatbot: Maaf, saya tidak dapat menemukan jawaban yang relevan.")
    else:
        print(f"Chatbot: Pertanyaan yang paling relevan: {questions[top_index]}")

# Meminta input dari pengguna
user_input = input("Ketik pertanyaan Anda: ")
handle_user_query(user_input)
```

```python
def chatbot():
    salam = "Halo! Bagaimana saya bisa membantu Anda hari ini?"
    print(salam)

    while True:
        input_pengguna = input("Anda: ").lower()
        
        if input_pengguna == "keluar":
            print("Selamat tinggal!")
            break
        elif input_pengguna == "hai" or input_pengguna == "halo":
            print("Chatbot: Halo! Senang bertemu dengan Anda.")
        elif input_pengguna == "apa kabar?":
            print("Chatbot: Saya baik-baik saja, terima kasih!")
        elif input_pengguna == "siapa nama kamu?":
            print("Chatbot: Nama saya adalah Chatbot.")
        else:
            print("Chatbot: Maaf, saya tidak mengerti apa yang Anda katakan.")

chatbot()
```

```python
import pandas as pd

data = pd.read_csv('faq_dataset.csv')
data.head()
```

```python
import re
from nltk.corpus import stopwords
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

# Inisialisasi stemmer
factory = StemmerFactory()
stemmer = factory.create_stemmer()

# Preprocessing function
def preprocess(text):
    text = text.lower()
    text = re.sub(r'\d+', '', text)
    text = re.sub(r'[^\w\s]', '', text)
    words = text.split()
    stop_words = set(stopwords.words('indonesian'))
    words = [word for word in words if word not in stop_words]
    words = [stemmer.stem(word) for word in words]
    return ' '.join(words)

# Terapkan preprocessing pada dataset
data['pertanyaan_clean'] = data['Pertanyaan'].apply(preprocess)

# Menampilkan hasil preprocessing
print(data[['Pertanyaan', 'pertanyaan_clean']].head())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Inisialisasi model TF-IDF
vectorizer = TfidfVectorizer()

# Terapkan TF-IDF ke pertanyaan yang sudah diproses
tfidf_matrix = vectorizer.fit_transform(data['pertanyaan_clean'])

# Menampilkan fitur kata-kata yang digunakan oleh TF-IDF
print(vectorizer.get_feature_names_out())
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# Input query dari pengguna
query = "Bagaimana cara memperbaiki koneksi wifi?"
query_processed = preprocess(query)

# Hitung TF-IDF untuk query pengguna
query_tfidf = vectorizer.transform([query_processed])

# Hitung kemiripan antara query dan pertanyaan di dataset
similarity_scores = cosine_similarity(query_tfidf, tfidf_matrix)

# Temukan pertanyaan dengan skor kemiripan tertinggi
ranked_indices = similarity_scores.argsort()[0][::-1]
top_index = ranked_indices[0]

# Tampilkan pertanyaan dan jawaban yang paling relevan
print(f"Pertanyaan: {data['Pertanyaan'][top_index]}")
print(f"Jawaban: {data['Jawaban'][top_index]}")
```

```python
# Fungsi chatbot untuk menangani input pengguna dan memberikan jawaban
def chatbot():
    while True:
        # Meminta input dari pengguna
        query = input("Anda: ")
        
        # Jika pengguna mengetik 'keluar', chatbot akan berhenti
        if query.lower() == 'keluar':
            print("Chatbot: Terima kasih! Sampai jumpa lagi.")
            break
        
        # Preprocessing query dari pengguna
        query_processed = preprocess(query)
        
        # Hitung TF-IDF untuk query pengguna
        query_tfidf = vectorizer.transform([query_processed])
        
        # Hitung cosine similarity antara query dan pertanyaan di dataset
        similarity_scores = cosine_similarity(query_tfidf, tfidf_matrix)
        
        # Temukan pertanyaan dengan skor kemiripan tertinggi
        ranked_indices = similarity_scores.argsort()[0][::-1]
        top_index = ranked_indices[0]
        
        # Tampilkan pertanyaan dan jawaban yang paling relevan
        print(f"Chatbot: {data['Jawaban'][top_index]}")

# Memulai chatbot
print("Chatbot: Halo! Ketik pertanyaan Anda atau ketik 'keluar' untuk mengakhiri obrolan.")
chatbot()
```
