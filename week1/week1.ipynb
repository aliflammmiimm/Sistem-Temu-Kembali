```python
import csv

with open('contoh_tulis.csv', 'w', newline = '') as file:
    writer = csv.writer(file)
    writer.writerow(['Nama', 'Usia', 'Kota'])
    writer.writerow(['Budi', '25', 'Jakarta'])
    writer.writerow(['Ali', '20', 'Bandung'])
```

```python
def cek_genap_ganjil(bilbul):
    if bilbul % 2 == 0:
        return "Genap"
    else:
        return "Ganjil" 

print(cek_genap_ganjil(17))
print(cek_genap_ganjil(26))
```
